Fill the missing values using mean

```python
for col in df.columns:

    if df[col].isnull().sum() > 0:

        if pd.api.types.is_numeric_dtype(df[col]):
            # Numerical columns: Mean
            if df[col].notna().any():
                df[col] = df[col].fillna(df[col].mean())
            else:
                df[col] = df[col].fillna(0)

        else:
            # Text columns: Mode
            if not df[col].mode().empty:
                df[col] = df[col].fillna(df[col].mode()[0])
            else:
                df[col] = df[col].fillna("Unknown")
```

Check Missing Values after Filling

```python
print("Missing values after filling:")
print(df.isnull().sum())

print("\nTotal missing values:",
      df.isnull().sum().sum())
```

Data Visualisation
A. Histogram – Numerical data distribution

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_performance_eda_dataset_50_rows.csv")

num_cols = df.select_dtypes(include=np.number).columns

df[num_cols].hist(figsize=(12, 8), bins=10, edgecolor="black")

plt.suptitle("Student Performance - Histograms")
plt.tight_layout()
plt.show()
```

boxplot

```python
num_cols = ["Age", "Attendance_Percent",
            "Study_Hours_Per_Day", "Marks"]

for col in num_cols:
    plt.figure(figsize=(7, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

count plot

```python
cat_cols = ["Gender", "Department", "Result"]

for col in cat_cols:
    plt.figure(figsize=(7, 4))
    sns.countplot(data=df, x=col)
    plt.title(f"Countplot of {col}")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

Correlation Heatmap

```python
num_cols = ["Age", "Attendance_Percent",
            "Study_Hours_Per_Day", "Marks"]

plt.figure(figsize=(8, 6))

sns.heatmap(
    df[num_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance - Correlation Heatmap")
plt.tight_layout()
plt.show()
```

scatter plot

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours_Per_Day",
    y="Marks",
    hue="Result",
    s=80
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()
```
